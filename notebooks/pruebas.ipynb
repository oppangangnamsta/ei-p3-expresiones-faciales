# Procesamiento de datos

```python
aux_img = np.random.choice(os.listdir("data/img_align_celeba"))
img = cv2.imread(f"data/img_align_celeba/{aux_img}", cv2.IMREAD_COLOR_RGB)

csv = pd.read_csv("data/list_landmarks_align_celeba.csv")
csv_para_img = csv[csv["image_id"] == aux_img]

result = model.predict(f"data/img_align_celeba/{aux_img}")[0].boxes.data.cpu()[:, :-2]  # type: ignore
corners = result

# img = cv2.imread("Imagen pegada.png", cv2.IMREAD_COLOR_RGB)
fig, ax = plt.subplots(figsize=(20, 10))
ax.imshow(img)  # type: ignore
esquinas = corners[:, :4]  # type: ignore
for puntos in esquinas:
    aux = puntos.reshape(2, 2)
    ancho = aux[0, 0] - aux[1, 0]
    alto = aux[0, 1] - aux[1, 1]
    rectangulo = Rectangle(
        (aux[1, 0], aux[1, 1]),  # type: ignore
        ancho,  # type: ignore
        alto,  # type: ignore
        edgecolor="red",  # Color del borde
        facecolor="none",  # 'none' para que sea transparente por dentro
        linewidth=1,  # Grosor de la línea
    )
    ax.add_patch(rectangulo)
plt.scatter(csv_para_img["lefteye_x"], csv_para_img["lefteye_y"])
plt.show()
```
